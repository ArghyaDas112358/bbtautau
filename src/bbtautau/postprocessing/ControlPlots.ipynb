```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parameters (NNLO+NNLL 13.6 TeV) ---
c2, c1, c0 = 0.263, -1.310, 2.047
sigma_SM = 31.05  # fb  (SM HH cross section at 13.6 TeV)

# --- κλ grid and σ/σSM curve ---
kappa = np.linspace(-10, 10, 400)
sigma_ratio = c2*kappa**2 + c1*kappa + c0

# --- Your expected Run-3 sensitivity ---
limit_ratio = 1.0       # expected 95% CL limit ≈ 1×SM
obs_limit_ratio = 1.2    # example observed (optional)

# --- Plot ---
plt.figure(figsize=(8,5))
plt.plot(kappa, sigma_ratio, 'b', lw=2.5, label=r'$\sigma_{HH}(\kappa_\lambda)/\sigma_{SM}$ (NNLO)')
plt.axhline(limit_ratio, color='r', ls='--', lw=2, label='Expected Run-3 limit')
plt.axhline(obs_limit_ratio, color='darkred', ls=':', lw=1.8, label='Observed limit')
plt.axvline(1, color='k', lw=1, ls=':', alpha=0.6)

# Highlight excluded regions (σ>limit)
plt.fill_between(kappa, sigma_ratio, limit_ratio,
                 where=(sigma_ratio>limit_ratio),
                 color='red', alpha=0.15)

plt.yscale('log')
plt.ylim(0.1, 100)
plt.xlabel(r'$\kappa_\lambda = \lambda_{HHH}/\lambda_{HHH}^{SM}$', fontsize=13)
plt.ylabel(r'$\sigma_{HH}(\kappa_\lambda)/\sigma_{HH}^{SM}$', fontsize=13)
plt.title('Expected Run-3 sensitivity: Boosted HH → bb ττ', fontsize=14)
plt.legend()
plt.grid(True, which='both', ls='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

```python
MAIN_DIR = Path("../../../")
CHANNEL = CHANNELS["hh"]  # options: "hh", "he", "hm"

plot_dir = MAIN_DIR / f"plots/ControlPlots/25Apr17{CHANNEL.key}"
plot_dir.mkdir(parents=True, exist_ok=True)

year = "2022"

base_dir = Path("/ceph/cms/store/user/rkansal/bbtautau/skimmer/")
data_paths = {
    "signal": base_dir / "25Apr17bbpresel_v12_private_signal",
    "data": base_dir / "25Apr17bbpresel_v12_private_signal",
    "bg": base_dir / "25Apr17bbpresel_v12_private_signal",
}

sigs = {s + CHANNEL.key: SAMPLES[s + CHANNEL.key] for s in SIGNALS}
bgs = {bkey: b for bkey, b in SAMPLES.items() if b.get_type() == "bg"}
```
