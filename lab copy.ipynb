```python
import json
import os
import pathlib
import re

# Directory containing the JSON files
data_directory = pathlib.Path("data/")

# Find all JSON files in the directory
json_files = sorted([f for f in os.listdir(data_directory) if f.endswith(".json")])

if not json_files:
    print(f"No JSON files found in '{data_directory}'")

# Loop through each JSON file
for file_name in json_files:
    print(f"File: {file_name}")
    file_path = data_directory / file_name
    
    # Keep track of unique versions found in this file
    versions_found = set()

    try:
        with open(file_path, 'r') as f:
            data = json.load(f)

        # Look for file paths in the JSON data
        for sample, subsamples in data.items():
            if isinstance(subsamples, dict):
                for subsample, file_list in subsamples.items():
                    if isinstance(file_list, list):
                        # Use regex to find the NanoAOD version in each path
                        for path in file_list:
                            match = re.search(r"NanoAODv\d+", path)
                            if match:
                                versions_found.add(match.group(0))

    except (json.JSONDecodeError, KeyError, TypeError) as e:
        print(f"  Could not process file {file_name}. Error: {e}")
        continue

    # Print the results for the current file
    if versions_found:
        print(f"NanoAOD versions: {', '.join(sorted(list(versions_found)))}")
    else:
        print(f"  No NanoAOD versions found in this file.")
    print()
```

```python
import sys, subprocess, json, pathlib, os, shutil
from datetime import datetime
now = datetime.now()
timestamp = now.strftime("%Y%m%d_%H%M%S")

# pick where results go (EOS or scratch)
OUTDIR = f"/eos/uscms/store/user/{os.getenv('USER')}/bbtautau/skims"
OUTDIR += "/HHbbtt_2022_" + timestamp
os.makedirs(OUTDIR, exist_ok=True)

# wipe old test symlinks if any, then point skimmer dirs to EOS
for d in ("outparquet", "outfiles"):
    if os.path.islink(d) or os.path.exists(d):
        shutil.rmtree(d, ignore_errors=True)
    os.symlink(OUTDIR, d)

# ╭──────────────── option A – ONE subsample ─────────────╮
cmd_one = [
    sys.executable, "src/run.py",
    "--samples", "HHbbtt",
    "--subsamples", "GluGlutoHHto2B2Tau_kl-1p00_kt-1p00_c2-0p00",
    "--starti", "0", "--endi", "1",
    "--year", "2022", "--processor", "skimmer",
    "--fatjet-pt-cut", "0", "--file-tag", "test"
]

idx = json.loads(pathlib.Path("data/index_2022.json").read_text())
subsamples = list(idx["HHbbtt"])
cmd_all = [
    sys.executable, "src/run.py",
    "--samples", "HHbbtt",
    *sum([["--subsamples", s] for s in subsamples], []),
    "--starti", "0", "--endi", "99999",
    "--year", "2022", "--processor", "skimmer",
    "--fatjet-pt-cut", "0", "--file-tag", "full"
]

```

```python
# ───── uncomment ONE of the two lines below to launch ─────
subprocess.run(cmd_one, check=True)   # single subsample
```

```python
# final wrapper done: run this in the morning
import sys, subprocess, json, pathlib, os, shutil
from datetime import datetime

now      = datetime.now()
stamp    = now.strftime("%Y%m%d_%H%M%S")

user     = os.getenv("USER", "unknown")
eos_base = f"/eos/uscms/store/user/{user}/bbtautau/skims"
dest     = f"{eos_base}/HHbbtt_2022_{stamp}"
os.makedirs(dest, exist_ok=True)
print(f"[{stamp}] outputs will move → {dest}")

idx      = json.loads(pathlib.Path("data/index_2022.json")
                      .read_text())
subs     = list(idx["HHbbtt"])
cmd      = [sys.executable, "src/run.py",
            "--samples", "HHbbtt",
            *sum([["--subsamples", s] for s in subs], []),
            "--starti", "0", "--endi", "999999",
            "--year", "2022", "--processor", "skimmer",
            "--fatjet-pt-cut", "0",
            "--file-tag", f"full_{stamp}",
            "--batch-size", "50"]         

print(f"[{datetime.now():%Y%m%d_%H%M%S}] launching skimmer")
subprocess.run(cmd, check=True)
print(f"[{datetime.now():%Y%m%d_%H%M%S}] skimmer done")

# move local outputs into EOS (May be we should add it in run.py)
for d in ("outparquet", "outfiles"):
    if os.path.exists(d):
        shutil.move(d, f"{dest}/{d}")
        print(f"moved {d} → {dest}/{d}")

print(f"[{datetime.now():%Y%m%d_%H%M%S}] ALL DONE")
```
